```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('penguins')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
column = ['bill_length_mm','bill_depth_mm','flipper_length_mm','body_mass_g']
for col in column:
  mean_value = df[col].mean()
  df[col] = df[col].fillna(mean_value)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['sex'].isnull().sum()
```

```python
mode_value = df['sex'].mode()[0]
df['sex'] = df['sex'].fillna(mode_value)
# mode_value
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['sex'] = df['sex'].map({'Male':1,'Female':0})
```

```python
df['island'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['island'], drop_first=True ,dtype=int)
```

```python
df.head()
```

```python
x = df.drop(['species'],axis=1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

model_lo = LogisticRegression()
model_lo.fit(x_train_scaled, y_train)
```

```python
y_pred_lo = model_lo.predict(x_test_scaled)
y_pred_lo
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy_score(y_pred_lo, y_test)
```

```python
print("Confusion Matrix: \n", confusion_matrix(y_pred_lo,y_test))
print("\nClassification Report: \n", classification_report(y_pred_lo, y_test))
```

```python
import numpy as np

def check_is_penguins():
  print("--------------------Check Is Penguin Species--------------------")

  try:
    bill_length_mm = float(input('Enter penguin bill length in (mm): '))
    bill_depth_mm = float(input('Enter penguin bill depth in (mm): '))
    flipper_length_mm = float(input('Enter penguin flipper length in (mm): '))
    body_mass_g = float(input('Enter penguin body mass in (g): '))
    sex_input = input('Enter sex(Male/Female): ').lower()
    island_input = input("Enter island(Dream, Torgersen, Biscoe): ").lower()
  except ValueError:
    print("\nInvalid input. Please enter appropriate values.")
    return

  #prepare features
  sex = 1 if sex_input == 'male' else 0
  island_Dream = 1 if island_input == 'dream' else 0
  island_Torgersen = 1 if island_input == 'torgersen' else 0

  feature_df = pd.DataFrame([[
      bill_length_mm,
      bill_depth_mm,
      flipper_length_mm,
      body_mass_g,
      sex,
      island_Dream,
      island_Torgersen
  ]], columns=x_train.columns) #Use x_train.columns to ensure order is correct

  scaled_feature = scaler.transform(feature_df)

  predicted_species = model_lo.predict(scaled_feature)
  prediction_proba = model_lo.predict_proba(scaled_feature)


  print("\n----------------------------------")
  print(f"Predicted Penguin Species: {predicted_species[0]}")
  print(f"Confidence: {np.max(prediction_proba) * 100:.2f}%")
  print("----------------------------------")

check_is_penguins()
```
