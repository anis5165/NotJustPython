```python
import cv2
```

```python
cap = cv2.VideoCapture(0) #0 - for default webcam, 1 - for external webcam and "video_file.mp4"

while True:
    #ret: a boolean (True if a frame is successfully read, False otherwise)
    #frame: the actual image (as a NumPy array)
    ret, frame = cap.read() # ret=True/False, frame=the image from webcam

    #If ret is False, something went wrong (e.g., webcam not available, disconnected).
    if not ret:
        print("Could not read frame")
        break
    cv2.imshow("Webcam feed",frame)


    #cv2.waitKey(1) waits 1 millisecond for a key press.
    #& 0xFF makes sure it works properly across all systems.
    #ord('q') is the ASCII code for 'q'.
    # So this condition means: if the user presses the 'q' key, exit the loop.
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quitting...")
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

#Defines the codec (compressor–decompressor) used to save the video.
#'XVID' is a popular codec that creates .avi files.
codec = cv2.VideoWriter_fourcc(*'XVID')

#Creates a VideoWriter object to save video.
'''"
my_video.avi" → output filename.
codec → compression format (XVID).
20 → frames per second (fps).
(frame_width, frame_height) → resolution.
'''
recorded = cv2.VideoWriter("my_video.avi", codec, 20,(frame_width, frame_height))

while True:
    success, image = cap.read()

    if not success:
        break

    recorded.write(image)
    cv2.imshow("Recording live",image)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
recorded.release()
cv2.destroyAllWindows()
```
