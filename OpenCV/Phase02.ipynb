```python
import cv2
```

```python
#resize image

image = cv2.imread("img.png")

if image is None:
    print("Image is not found")
else:
    print("Image loaded")

    resized = cv2.resize(image, (300,300)) # "(width, height)"
    cv2.imshow("original image",image)
    cv2.imshow("resized image",resized)
    cv2.imwrite("resized-output.png",resized)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#cropped

image = cv2.imread("img.png")

if image is None:
    print("Image is not found")
else:
    print("Image loaded")

    cropped = image[0:500 , 0:350]  #[startY:endY , startX:endX]  coordinates 
    cv2.imshow("cropped image",cropped)
    cv2.imshow("original image",image)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# image rotation
# M = cv2.getRotationMatrix2D(center, angle, scale)
# rotated_image = cv2.warpAffine(image, M, (Width, height))

image = cv2.imread("KAAM-WALI.jpg")

if image is None:
    print("Image is not found")
else:
    (h, w) = image.shape[:2]

    center = (w//2, h//2)
    M = cv2.getRotationMatrix2D(center, 90, 2.0) #1.0 - same size, 0.5 - half size, 2.0 - double size
    rotated_image = cv2.warpAffine(image, M, (w,h))

    cv2.imshow("original image", image)
    cv2.imshow("rotated image", rotated_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# image flip 
# fliped_image = cv2.filp(image, flipcode) 0 - vertically(top to bottom), 1 - horizontally(left to right), -1 - both H and V

image = cv2.imread("resized-output.png")

if image is None:
    print("Image is not found")
else:
    fliped_horizontal = cv2.flip(image,1)
    fliped_vertical = cv2.flip(image,0)
    fliped_both = cv2.flip(image,-1)
    
    cv2.imshow("Horizontal",fliped_horizontal)
    cv2.imshow("vertical",fliped_vertical)
    cv2.imshow("Both flip",fliped_both)
    cv2.imshow("original",image)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
