```python
import cv2
```

```python
image = cv2.imread("./img.png")

if image is not None:
    cv2.imshow("Auto Docs",image) #open the window
    cv2.waitKey(0) #jab tak keyboard se koi key nhi dabai jati tab tak window ko open rkho
    cv2.destroyAllWindows() #close the window
    
else:
    print("Could not load the image")
```

```python
image = cv2.imread("./img.png")

if image is not None:
    success = cv2.imwrite("auto-docs.png",image)
    if success:
        print("Image save successfully as 'auto-docs.png'")
    else:
        print("Failed to save an image")
else:
    print("Error: Could not load image")
```

```python
#dimension

image = cv2.imread("./img.png")

if image is not None:
    h, w, c = image.shape
    print(f"Image Loaded:\nHeight: {h}\nWidth: {w}\nChannels: {c}")
else:
    print("Error: Could not load image")
```

```python
#grey scale conversion

image = cv2.imread("./img.png")

if image is not None:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    cv2.imshow("Grayscale Image", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows
else:
    print("Error: Could not load image")
```
