```python
import cv2
```

```python
#line draw
image = cv2.imread("resized-output.png")

if image is None:
    print("Oops! Your image is not working")
else:
    print("Image loaded successfully!")

    pt1 = (50,100)
    pt2 = (200,200)
    color = (0, 0, 255)
    cv2.line(image, pt1, pt2, color, thickness=4)

    cv2.imshow("Line Drawing", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# draw rectangle
image = cv2.imread("resized-output.png")

if image is None:
    print("Oops! Your image is not working")
else:
    print("Image loaded successfully!")

    pt1 = (50,50)
    pt2 = (130,150)
    color = (0, 255, 0)
    cv2.rectangle(image, pt1, pt2, color, thickness=2) #in rectangle pt1 - top left cornor, pt2 - bottom right cornor

    cv2.imshow("Line Drawing", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# draw circle 

image = cv2.imread("resized-output.png")

if image is None:
    print("Oops! Your image is not working")
else:
    print("Image loaded successfully!")

    center = (150,150)
    radius = 50
    color = (0, 255, 0)
    cv2.circle(image, center, radius, color, thickness=-1) #-1 - fill color in circle and in +value use as thickness 

    cv2.imshow("Line circle", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#Adding text on image

image = cv2.imread("resized-output.png")

if image is None:
    print("Oops! Your image is not working")
else:
    print("Image loaded successfully!")

    org = (30,150)
    color = (255, 255, 0)
    cv2.putText(image, "Hii!! kaise ho...", org, cv2.FONT_HERSHEY_COMPLEX_SMALL, 1.0, color, thickness=1) 

    cv2.imshow("Adding text on image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
